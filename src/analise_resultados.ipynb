```python
import csv
import plotly.graph_objects as go

dados_agrupados = {}

with open(r"C:\Users\vivif\OneDrive\Desktop\TrabalhosSillys\SEXTO SEMESTRE\Computação Paralela\Trabalho AV2\Ordenacoes\src\resultados_ordenacao.csv", mode='r', encoding='utf-8') as arquivo:
    leitor = csv.reader(arquivo, delimiter=';')

    for linha in leitor:
        if len(linha) < 5 or not linha[2].isdigit():
            continue

        algoritmo = linha[1]
        tamanho = int(linha[2])
        serial = int(linha[3])
        paralelo = int(linha[4].replace('\\', ''))

        chave = (algoritmo, tamanho)
        if chave not in dados_agrupados:
            dados_agrupados[chave] = {'seriais': [], 'paralelos': []}

        dados_agrupados[chave]['seriais'].append(serial)
        dados_agrupados[chave]['paralelos'].append(paralelo)


cor_serial = "#fff1a1"
cor_paralelo = "#ffa6bc"

for (algoritmo, tamanho), tempos in dados_agrupados.items():
    tamanho_str = "100k" if tamanho == 100000 else "10M"
    
    seriais = tempos['seriais']
    paralelos = tempos['paralelos']
    
    rotulos_testes = [f"Teste {i+1}" for i in range(len(seriais))]

    fig = go.Figure()

    fig.add_trace(go.Bar(
        x=rotulos_testes,
        y=seriais,
        name='Serial',
        marker_color=cor_serial,
        text=[f"{v} ms" for v in seriais],
        textposition='auto'
    ))

    fig.add_trace(go.Bar(
        x=rotulos_testes,
        y=paralelos,
        name='Paralelo',
        marker_color=cor_paralelo,
        text=[f"{v} ms" for v in paralelos],
        textposition='auto'
    ))

    fig.update_layout(
        title=f'Comportamento em Execuções Individuais: {algoritmo} ({tamanho_str})',
        xaxis_title='Número do Teste',
        yaxis_title='Tempo de Execução (ms)',
        barmode='group',
        template='plotly_dark',
        title_x=0.5,
        width=800,
        height=500
    )

    fig.show()
```
